```python
rename_columns = {
    "ID & Farm Name": "id_farm_name",
    "Date": "date",
    "Vendor ID": "vendor_id",
    "Activity ID": "activity_id",
    "Crop ID": "crop_id",
    "Crop Name": "crop_name",
    "Water Used (L)": "water_used_l",
    "Nutrient Used (ml)": "nutrient_used_ml",
    "Energy Used (kWh)": "energy_used_kwh",
    "Temperature (°C)": "temperature_c",
    "Humidity (%)": "humidity_pct",
    "pH": "ph",
    "EC (mS/cm)": "ec_ms_cm",
    "Yield (kg)": "yield_kg",
    "Labor Hours": "labor_hours",
    "Status": "status",
    "Location": "location"
}
```

```python
farm_a = farm_a.rename(columns=rename_columns)
farm_b = farm_b.rename(columns=rename_columns)
farm_c = farm_c.rename(columns=rename_columns)
```

```python
farm_a.columns
```

```python
def split_id_farm(df):
    
    split_data = df["id_farm_name"].astype(str).str.strip().str.split(
        n=1,
        expand=True
    )
    
    df["record_id"] = split_data[0].str.upper()
    df["farm_name"] = split_data[1].str.title()
    
    df = df.drop(columns=["id_farm_name"])
    
    return df
```

```python
farm_a = split_id_farm(farm_a)
farm_b = split_id_farm(farm_b)
farm_c = split_id_farm(farm_c)
```

```python
farm_a[["record_id", "farm_name"]].head()
```

```python
text_columns = [
    "vendor_id",
    "activity_id",
    "crop_id",
    "crop_name",
    "status",
    "location",
    "farm_name"
]

for df in [farm_a, farm_b, farm_c]:
    for col in text_columns:
        df[col] = df[col].astype(str).str.strip()
```

```python
for df in [farm_a, farm_b, farm_c]:
    
    df["vendor_id"] = df["vendor_id"].str.upper()
    df["activity_id"] = df["activity_id"].str.upper()
    df["crop_id"] = df["crop_id"].str.upper()
    
    df["crop_name"] = df["crop_name"].str.title()
    df["status"] = df["status"].str.title()
    df["location"] = df["location"].str.title()
```

```python
for df in [farm_a, farm_b, farm_c]:
    df["date"] = pd.to_datetime(df["date"], errors="coerce")
```

```python
farm_data["month"] = farm_data["date"].dt.to_period("M").astype(str)
```

```python
farm_data["year"] = farm_data["date"].dt.year
```

```python
farm_data["month_name"] = farm_data["date"].dt.month_name()
```

```python
farm_data["yield_per_100l_water"] = np.where(
    farm_data["water_used_l"] > 0,
    (farm_data["yield_kg"] / farm_data["water_used_l"]) * 100,
    np.nan
)
```

```python
farm_data["yield_per_kwh"] = np.where(
    farm_data["energy_used_kwh"] > 0,
    farm_data["yield_kg"] / farm_data["energy_used_kwh"],
    np.nan
)
```
